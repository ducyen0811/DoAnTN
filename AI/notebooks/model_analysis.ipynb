```python
import matplotlib.pyplot as plt
import seaborn as sns
display(data.isna().sum())
plt.figure(figsize=(11, 8))
sns.heatmap(data.corr(numeric_only=True), cmap='coolwarm', center=0)
plt.tight_layout()
plt.show()
```
